```python
# write a program to swap two number before swap
a=10
b=20
after swap
a=20
b=10

```

```python
# first method
a=10
b=20
print(f"before swap:\n a = {a}, b = {b}")
temp=a # temp is an empty jar we will store a value in that so a is empty
a=b    # a is empty right so we will push b value to a to swap so b is empty
b=temp  # b is empty right to swap we will take a value from temp which we stored earlier
print(f"before swap:\n a = {a}, b = {b}")
```

```python
# second method
a=10
b=20
print(f"before swap:\n a = {a}, b = {b}")
a,b = b,a
print(f"before swap:\n a = {a}, b = {b}")
```

```python
num = 535
temp=num
rev=0
while(num>0):
    rem=num%10
    rem=(rev*10)+rem
    num=num//10
print(f"Number:{temp}\nRevarsed num:{rev}")
if(temp==rev):
    print(temp,"is palindrome")
else:
    print(temp,"is not palindrome")
```
